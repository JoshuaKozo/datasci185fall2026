## <span style="color:darkblue"> I. Converting between types </span>

<font size="4">

Python gives us built-in functions to convert values deliberately: `str()` turns a value into text, while `int()` and `float()` turn text (or another number) into a number.

```python
# Three conversion functions:

count_as_text   = str(42)        # the string "42"
price_as_number = float("3.99")  # the float 3.99
year_as_number  = int("2024")    # the integer 2024

print(count_as_text,   type(count_as_text))
print(price_as_number, type(price_as_number))
print(year_as_number,  type(year_as_number))
```

```python
# WHY THIS MATTERS: arithmetic on strings is not what you expect.

a = "10"
b = "2"

print(a + b)            # "102"  <-- "+" CONCATENATES strings, it does not add
print(int(a) + int(b))  # 12     <-- convert first, then add

c = str((int(a) + int(b)))
print(c)
```

<font size = "4">

Sometimes the behavior of commands can be different when numbers are represented by strings vs. represented by integers or floats:

```python

# Month numbers input as strings
months_as_text = ["Nov", "Dec", "Jan", "Feb"]
sales          = [100, 250, 80, 300]

# Arranges the data in the same order as typed:
plt.bar(x = months_as_text, height = sales)
plt.title("x-values stored as strings")
plt.xlabel("month number (stored as text)")
plt.ylabel("sales")
plt.show()
```

```python
months_as_numbers = [11, 12, 1, 2]

plt.bar(x = months_as_numbers, height = sales)
plt.title("x-values stored as integers")
plt.xlabel("month number")
plt.ylabel("sales")
plt.show()
```

## <span style="color:darkblue"> III. Comparison operators </span>

- ### One of the most common ways Boolean objects show up is in the use of **comparison operators**.
- ### These are operators used to compare two values, and return either ``True`` or ``False``.
- ### There are six basic comparison operators. The first four are:
    - #### ``>`` - Greater than
    - #### ``<`` - Less than
    - #### ``>=`` - Greater than or equal to
    - #### ``<=`` - Less than or equal to

```python
# Test if one number is greater than the other
print(5 > 3)

print(3 > 9)

print(4 + 3 > 6)
```

```python
# Test if one number is less than the other
print(0 < -2.5)

print(1 < 1.01)

print(30/10 < 1)
```

```python
# Test if one number is greater that or equal to the other
print(10 >= 1)

print(8 >= 50)

print(2**3 >= 10)
```

```python
# Greater than OR equal to!
print(5 > 5) # this is false
print(5 >= 5) # this is true, since 5 equals 5
```

```python
# Test if one number is less than or equal to the other
print(10 <= 1)

print(8 <= 50)

print(2**3 <= 10)
```

#### Two more basic comparison operators
- #### ``==`` - Equal
- #### ``!=`` - Not equal

#### **Warning**: Do not confuse the assignment operator ``=`` with the comparison operator that tests for equality ``==``

```python
# Does 5*3 equal 15? (Yes)
print(5*3 == 15)

# Does 3 - 2 equal 7? (No)
print(3 - 2 == 7)
```

```python
# Is 5*3 + 1 different from 20? (yes)
print(5*3 + 1 != 20)

# Is 10/5 different from 2? (no)
print(10/5 != 2)
```

#### In most cases, we use a comparison operator with one or more variables 

```python
x = math.sin(4*math.pi/3)
print(x)
print(x > 0)

# are these the side lenths of a right triangle?
a = 3
b = 4
c = 5

# How should we test this?

print(a**2 + b**2 == c**2)
```

## <span style="color:darkblue"> IV. Strings and Lists </span>

- ##### The "equals to" comparison and "not equals to" comparison extend naturally to strings and lists
- ##### If every character in a string matches another string **exactly**, then they are equal.
- ##### If every element in a list matches **every** element in another list, then they are equal. **Order of the elements matters**

```python
x = "Hello World!"
y = "Hello World !"

print(x == "Hello World!")
print(y == "Hello World!")

print(x != y)

print('ab' == 'ab')
```

```python
list_1 = ["blue", "red", "green"]
list_2 = ["blue", "red", "green"]
list_3 = ["red", "green", "blue"]
list_4 = ["red", "green", "blue", "violet"]

print(list_1 == list_2)
print(list_1 == list_3) # False: different ordering
print(list_3 == list_4) # False: different number of elements
print(list_3 != list_4) # True
```

#### The other comparison operators "<", ">", "<=", and ">=" can also be used for strings and lists. However, the behavior is not obvious. See the exercises for more.

## <span style="color:darkblue"> V. The "in" keyword </span>

#### This keyword allows us to test whether
- #### A character appears in a string
- #### An object appears in a list
- #### A number appears in a numpy array

```python
# The first way to use the "in" command is to check whether a word is contained 
# in a sentence. This can be useful if you're trying to search for patterns

keyword = "economic"
sentence = "The Federal Reserve makes forecasts about many economic outcomes"

print(keyword in sentence)

keyword = "bank"
print(keyword in sentence)

search_word = "Federal" 
print(search_word in sentence)

important_word = "federal"
print(important_word in sentence)
```

```python
# Use the condition "in" to test whether a word is part of a list
# when the right-hand size is a list "word in list" will search for an exact match

current_month  = "September"

list_summer_months = ["June","July","August"]

print(current_month in list_summer_months)
print('June' in list_summer_months)
```

```python
# Use the keyword "in" to see if a number is in a list
# (Later we will see that "in" also works for numpy arrays)

s = [1.0, 3.0, -17, 12]
print(1 in s)
print(type(int(1)) == type(s[1]))
```

##  <span style="color:darkblue"> VI. Logical Operators </span>

- #### The four logical operators are "not", "and", "or", "^" (this last one is exclusive or)

```python
p = True
q = False

print(not p)
print(not q)
```

```python
age  = 22

# True or false? This person can vote in the U.S.

# 1st way to test it:
print(age >= 18)
```

```python
age  = 22

# True or false? This person can vote in the U.S.

# Do the same test using the "not" keyword

print(not (age < 18))
```

```python
p = True
q = False
a = True
b = False

print(p and a)
print(p and b)
print(q and a)
print(q and b)
```

```python
price = 24.5

# True or false? The price is greater than 10 dollars 
# and less than or equal to 20 dollars.

print(price > 10 and price <= 20)
```

```python
# can assign Boolean objects to variables

price = 9

statement_1 = price > 10
statement_2 = price <= 20

print(statement_1)
print(statement_2)
print(statement_1 and statement_2)
```

<font size = "4">

This also works:

```python
print(10 < price <= 20)
```

#### **Both** statements must be true for [statement_1] and [statement_2] to be true

#### To test if **at least** one statment is true, use the "or" operator


<font size = "4">

The standard "or" is **inclusive or**

```python
p = True
q = False
a = True
b = False

print(p or a)
print(p or b)
print(q or a)
print(q or b)
```

```python
st_1 = 5 > 3
st_2 = 1 >= 2000

print(st_1 or st_2)
print()

st_3 = 9 < 19
st_4 = 0 < 1
print(st_3 or st_4)
print()

st_5 = -1 >= 10
st_6 = 12 > 13
print(st_5 or st_6)
print()
```

#### To test if **exactly** one statment is true, use the "^" operator

<font size = "4">

This is **exclusive or**

```python
p = True
q = False
a = True
b = False

print(p ^ a)
print(p ^ b)
print(q ^ a)
print(q ^ b)
```

#### While not necessary, can separate statements with parentheses for readibility

```python
x = 13
y = -5
z = False

P = (x > y) and (not z)
print(P)

Q = (x == 5) or (y > 10)
print(Q)
```
